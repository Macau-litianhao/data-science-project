## Executive Summary

Between 2000 and 2021, China's **age-standardised stroke death rate fell by 37.2%**, from 187.1 to 117.5 deaths per 100,000. Over the same period, **total stroke deaths rose by 33.6%**, from about 1.89 million to 2.53 million. The crude rate, calculated from actual deaths and actual population, rose by 18.9%.

The apparent contradiction is not a data error. The two headline measures answer different questions:

- The age-standardised rate approximates **individual risk under a fixed age structure**.
- Total deaths measure the **actual burden across the real population**.

The main demographic mechanism is population ageing. The population aged 65+ grew by **111.4%**, while the total population grew by **12.4%**. Stroke deaths grew by 33.6%, which is above total population growth but below growth of the 65+ group. This pattern is consistent with a genuine fall in age-specific risk and a simultaneous increase in the number of older people.

> **Interpretation limit.** The notebook establishes the direction of the effects, but it does not produce a formal decomposition. A precise statement such as "X% of the increase is caused by ageing" would require age-specific stroke death counts, which are not present in these files.

A stacked or 100% chart would remove the absolute population growth that is central to the explanation. Ordering the groups from oldest to youngest directs attention from the fastest-growing group to the groups that declined.

# About the Datasets

## Three concepts used in the analysis

**Concept 1: age-standardised rate versus crude rate.**

- **Crude rate** = deaths / actual population x 100,000. It describes what happened to the real population.
- **Age-standardised rate** = deaths / a hypothetical population with a fixed age structure x 100,000. It compares risk while holding age structure constant.

**Concept 2: aggregation level.**

The stroke columns are already aggregated over sex and age. The notebook therefore cannot split the results by sex or calculate age-specific death rates. This limitation matters for the interpretation.

**Concept 3: country-year joins.**

`Entity` is the readable name, `Code` is the ISO country code, `Year` is the time period, and the remaining columns contain values. Tables are joined on `Code` and `Year`, not on `Entity`, because display names can differ across files.

## Coverage note

The two stroke files cover 2000-2021 and align exactly. The population files cover a wider period, so the notebook filters them to the analysis window.

# Module 2: Data Wrangling


### 2.1 Remove non-country rows

Our World in Data files include regional and global aggregates such as `World`, `Africa`, and `Asia`. Some use codes beginning with `OWID_` or `UN_`. Keeping a row only when its code is a three-letter ISO code prevents these aggregates from entering the analysis.

The filter is applied consistently to all four datasets.


```python
aggregates = stroke_rate[
    stroke_rate["Code"].fillna("").astype(str).str.startswith(("OWID_", "UN_"))
]

print("Number of aggregate rows:", len(aggregates))
print("Aggregate entities:", sorted(aggregates["Entity"].unique()))
```

```python
def keep_real_countries(df):
    """Keep rows whose Code is a three-letter ISO country code."""
    code_str = df["Code"].fillna("").astype(str).str.upper()
    is_iso3 = code_str.str.fullmatch(r"[A-Z]{3}")
    is_not_aggregate = ~code_str.str.startswith(("OWID_", "UN_"))
    return df[is_iso3 & is_not_aggregate].copy()


stroke_countries = keep_real_countries(stroke_rate)
deaths_countries = keep_real_countries(stroke_deaths)
pop_countries = keep_real_countries(population)
pop_age_countries = keep_real_countries(pop_age)

for name, df in {
    "stroke_rate": stroke_rate,
    "stroke_deaths": stroke_deaths,
    "population": population,
    "pop_age": pop_age,
}.items():
    filtered = keep_real_countries(df)
    print(f"{name:<15} {len(df):>6,} rows -> {len(filtered):>6,} real-country rows")

print()
print("Countries in the stroke-rate table:", stroke_countries["Entity"].nunique())
print("Year range:", stroke_countries["Year"].min(), "-", stroke_countries["Year"].max())
```

**How the filter works.**

- `fillna("").astype(str)` prevents missing codes from being mistaken for valid identifiers.
- `.str.fullmatch(r"[A-Z]{3}")` keeps exactly three-letter ISO codes.
- `~code_str.str.startswith(("OWID_", "UN_"))` excludes aggregate prefixes.
- The same function is reused for all four files.


### 2.2 Check for missing values

Missing values are checked after filtering so that the validation reflects the data actually used.


```python
missing_summary = pd.DataFrame({
    "Dataset": ["stroke_rate", "stroke_deaths", "population", "pop_age"],
    "Rows": [
        len(stroke_countries),
        len(deaths_countries),
        len(pop_countries),
        len(pop_age_countries),
    ],
    "Missing values": [
        int(stroke_countries.isna().sum().sum()),
        int(deaths_countries.isna().sum().sum()),
        int(pop_countries.isna().sum().sum()),
        int(pop_age_countries.isna().sum().sum()),
    ],
})

print(missing_summary.to_string(index=False))
```

### 2.3 Merge the two core stroke tables

The rate and death count are stored in separate files. They are joined on `Code` and `Year`, because each row represents one country-year. An inner join keeps only rows present in both files.


```python
stroke = pd.merge(
    stroke_countries[["Entity", "Code", "Year", "stroke_rate"]],
    deaths_countries[["Code", "Year", "stroke_deaths"]],
    on=["Code", "Year"],
    how="inner"
)

print("Rows in stroke-rate table   :", len(stroke_countries))
print("Rows in total-deaths table  :", len(deaths_countries))
print("Rows in merged stroke table :", len(stroke))
print("Missing values after merge  :", int(stroke.isna().sum().sum()))

assert len(stroke_countries) == len(deaths_countries) == len(stroke)
assert stroke.isna().sum().sum() == 0

stroke.head()
```

**Validation result.** The two stroke files cover exactly the same country-years. The inner join preserves every row and introduces no missing values.


### 2.4 Attach population and compute the crude rate

The crude rate uses China's real population in each year:

$$\text{Crude rate} = \frac{\text{Stroke deaths}}{\text{Population}} \times 100{,}000$$


```python
china_pop = pop_countries[
    (pop_countries["Entity"] == "China") &
    (pop_countries["Year"].between(2000, 2021))
][["Year", "Population"]].sort_values("Year")

assert len(china_pop) == 22
assert china_pop["Year"].tolist() == list(range(2000, 2022))

print(china_pop.head())
print("...")
print("Rows:", len(china_pop))
```

```python
china = stroke[stroke["Entity"] == "China"].sort_values("Year").copy()
china = pd.merge(china, china_pop, on="Year", how="inner", validate="one_to_one")
china["crude_rate"] = china["stroke_deaths"] / china["Population"] * 100_000
china["deaths_10k"] = china["stroke_deaths"] / 10_000

assert len(china) == 22
print(china[["Year", "stroke_rate", "stroke_deaths", "Population", "crude_rate"]].to_string(index=False))
```

### 2.5 Compare 2000 and 2021

The comparison table brings together the age-standardised rate, total deaths, population, and crude rate. Percentage changes show why the headline figures appear to move in opposite directions.


```python
row_2000 = china.loc[china["Year"] == 2000].iloc[0]
row_2021 = china.loc[china["Year"] == 2021].iloc[0]

summary = pd.DataFrame({
    "Indicator": [
        "Age-standardised stroke death rate (per 100,000)",
        "Total stroke deaths",
        "Total population",
        "Crude stroke death rate (per 100,000)",
    ],
    "Year 2000": [
        row_2000["stroke_rate"],
        row_2000["stroke_deaths"],
        row_2000["Population"],
        row_2000["crude_rate"],
    ],
    "Year 2021": [
        row_2021["stroke_rate"],
        row_2021["stroke_deaths"],
        row_2021["Population"],
        row_2021["crude_rate"],
    ],
})

summary["Change (%)"] = (
    (summary["Year 2021"] / summary["Year 2000"] - 1) * 100
).round(1)

display(summary.style.format({
    "Year 2000": "{:,.1f}",
    "Year 2021": "{:,.1f}",
    "Change (%)": "{:+.1f}%",
}))
```

**Main finding.** The age-standardised rate falls by **37.2%**, while total deaths rise by **33.6%**. The crude rate rises by **18.9%**, showing that the real population experienced a larger mortality burden even as age-adjusted risk declined.


### 2.6 Validate the arithmetic

The computed crude rate should reconstruct the reported death count when multiplied by population and divided by 100,000.


```python
check = china["crude_rate"] * china["Population"] / 100_000
difference = (check - china["stroke_deaths"]).abs()

assert np.allclose(check, china["stroke_deaths"], rtol=1e-12, atol=1e-6)

print("Largest difference:", difference.max())
print("Sanity check passed: the crude rate reconstructs total deaths.")
```

# Module 3: Exploratory Data Analysis


### 3.1 Chart 1: the paradox in one picture

Total deaths are plotted as bars on the left axis and the age-standardised rate as a line on the right axis. Two axes are necessary because deaths are measured in millions while the rate is measured per 100,000 people.


```python
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.bar(
    china["Year"],
    china["deaths_10k"],
    color="#F09595",
    edgecolor="#A32D2D",
    linewidth=0.6,
    label="Total stroke deaths (10,000)",
)
ax1.set_xlabel("Year")
ax1.set_ylabel("Total stroke deaths (10,000)", color="#A32D2D")
ax1.tick_params(axis="y", labelcolor="#A32D2D")
ax1.set_ylim(0, 300)

ax2 = ax1.twinx()
ax2.plot(
    china["Year"],
    china["stroke_rate"],
    color="#185FA5",
    linewidth=2.5,
    marker="o",
    markersize=3,
    label="Age-standardised death rate (per 100,000)",
)
ax2.set_ylabel("Age-standardised death rate (per 100,000)", color="#185FA5")
ax2.tick_params(axis="y", labelcolor="#185FA5")
ax2.set_ylim(100, 210)

ax1.set_title(
    "China: Total Stroke Deaths Rise While the Age-Standardised Rate Falls, 2000-2021",
    pad=14,
)
ax1.grid(axis="y", alpha=0.25)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(handles1 + handles2, labels1 + labels2, loc="upper center")

fig.tight_layout()
FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(exist_ok=True)
fig.savefig(FIGURE_DIR / "01_stroke_paradox.png", dpi=220, bbox_inches="tight")
plt.show()
```

**Reading Chart 1.**

- Total deaths increase from about 1.89 million to 2.53 million.
- The age-standardised rate falls from 187.1 to 117.5 per 100,000.
- The trends are not contradictory: one measures total burden, while the other measures risk under a fixed age structure.
- The rate does not fall in every year. It rises between 2000 and 2004 before turning downward. Measuring from the 2004 peak gives a larger decline of about 39.3%.


### 3.2 Chart 2: the ageing mechanism

Population ageing is the main demographic mechanism. The next table and chart compare the five age groups in 2000 and 2021.


```python
china_age = pop_age_countries[pop_age_countries["Entity"] == "China"].set_index("Year")
age_bands = ["Ages 65+", "Ages 25-64", "Ages 15-24", "Ages 5-14", "Under-5s"]

age_2000 = china_age.loc[2000, age_bands]
age_2021 = china_age.loc[2021, age_bands]

age_table = pd.DataFrame({
    "Population 2000": age_2000,
    "Population 2021": age_2021,
})
age_table["Change (%)"] = (
    (age_table["Population 2021"] / age_table["Population 2000"] - 1) * 100
).round(1)
age_table["Share 2000 (%)"] = (age_table["Population 2000"] / age_2000.sum() * 100).round(1)
age_table["Share 2021 (%)"] = (age_table["Population 2021"] / age_2021.sum() * 100).round(1)

display(age_table.style.format({
    "Population 2000": "{:,.0f}",
    "Population 2021": "{:,.0f}",
    "Change (%)": "{:+.1f}%",
    "Share 2000 (%)": "{:.1f}%",
    "Share 2021 (%)": "{:.1f}%",
}))

print("Total population change:",
      f"{(age_2021.sum() / age_2000.sum() - 1) * 100:+.1f}%")
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

x = np.arange(len(age_bands))
width = 0.38

ax.bar(
    x - width / 2,
    age_2000.values / 1e6,
    width,
    label="2000",
    color="#B5D4F4",
    edgecolor="#185FA5",
    linewidth=0.6,
)
ax.bar(
    x + width / 2,
    age_2021.values / 1e6,
    width,
    label="2021",
    color="#F0997B",
    edgecolor="#993C1D",
    linewidth=0.6,
)

ax.set_xticks(x, age_bands)
ax.set_xlabel("Age group")
ax.set_ylabel("Population (millions)")
ax.set_title("China's Population by Age Group, 2000 vs 2021", pad=14)
ax.legend()
ax.grid(axis="y", alpha=0.25)

fig.tight_layout()
fig.savefig(FIGURE_DIR / "02_ageing_mechanism.png", dpi=220, bbox_inches="tight")
plt.show()
```

**Reading Chart 2: the ageing mechanism.**

| Age group | 2000 to 2021 | Change |
|:--|--:|--:|
| Ages 65+ | 89.1 M to 188.3 M | **+111.4%** |
| Ages 25-64 | 666.1 M to 825.8 M | +24.0% |
| Ages 15-24 | 202.7 M to 161.5 M | -20.3% |
| Ages 5-14 | 226.4 M to 176.9 M | -21.9% |
| Under-5s | 85.2 M to 73.9 M | -13.3% |
| **Total** | 1,269.6 M to 1,426.4 M | **+12.4%** |

The 65+ population more than doubled, while every younger age band shrank. Because stroke mortality is concentrated at older ages, a much larger older population can produce more deaths even when age-specific risk declines.


### 3.3 Test whether ageing explains the entire rise

Three growth rates place the mortality increase in perspective: total deaths, total population, and the population aged 65+.


```python
deaths_growth = (row_2021["stroke_deaths"] / row_2000["stroke_deaths"] - 1) * 100
pop_growth = (row_2021["Population"] / row_2000["Population"] - 1) * 100
old_growth = (age_2021["Ages 65+"] / age_2000["Ages 65+"] - 1) * 100

print(f"Growth in total stroke deaths : {deaths_growth:+.1f}%")
print(f"Growth in total population    : {pop_growth:+.1f}%")
print(f"Growth in population aged 65+ : {old_growth:+.1f}%")
print()
print(
    "Deaths grew faster than total population, so population growth alone "
    "does not explain the increase."
)
print(
    "Deaths grew more slowly than the 65+ population, which is consistent "
    "with a fall in age-specific risk."
)
```

**Interpretation.** Deaths grew by 33.6%, which is above total population growth of 12.4% but below growth of the 65+ population of 111.4%. The two comparisons provide bounds: population growth alone under-predicts the increase, while one-for-one growth with the elderly population over-predicts it. The evidence is consistent with population ageing plus a genuine decline in age-specific risk.

> **Limitation.** This is an indirect comparison, not a formal demographic decomposition. A precise decomposition would require age-specific stroke death counts, which are not available in the four CSV files. The notebook should not claim a precise percentage caused by ageing.


### 3.4 What the analysis established

1. **Both headline trends are real.** The age-standardised rate falls while total deaths rise.
2. **The decline is not uniform.** The rate rises initially and peaks in 2004 before declining.
3. **Population ageing provides the main mechanism.** The 65+ group more than doubled.
4. **Ageing alone does not explain the full increase.** Deaths grow faster than total population but more slowly than the 65+ group, pointing to lower age-specific risk.
5. **The argument has a clear limit.** Without age-specific death counts, the notebook establishes direction rather than an exact decomposition.
